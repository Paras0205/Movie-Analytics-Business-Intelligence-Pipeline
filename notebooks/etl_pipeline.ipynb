```python
import requests
import pandas as pd
```

```python
from getpass import getpass

api_key = getpass("Enter your TMDB API key: ")
df_list = []

for i in range(1, 501):
    
    url = "https://api.themoviedb.org/3/discover/movie?api_key={}&language=en-US&page={}".format(api_key, i)
    
    response = requests.get(url)
    response.raise_for_status()

    temp_df = pd.DataFrame(response.json()['results'])

    df_list.append(temp_df)

df = pd.concat(df_list, ignore_index=True)
```

```python
from getpass import getpass

password = getpass("Enter MySQL Password: ")

conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=password,
    database="movie_analytics"
)
cursor = conn.cursor()
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS movies (
    id INT PRIMARY KEY,
    title VARCHAR(500),
    original_language VARCHAR(20),
    original_title VARCHAR(500),
    popularity DOUBLE,
    release_date DATE NULL,
    vote_average DOUBLE,
    vote_count INT,
    release_date_status VARCHAR(20),
    genre_ids TEXT
)
"""

cursor.execute(create_table_query)
conn.commit()

print("Movies table created")
```

```python
import json

df["genre_ids"] = df["genre_ids"].apply(json.dumps)
```

```python
print(df["genre_ids"].iloc[0])
print(type(df["genre_ids"].iloc[0]))
```

```python
data = df[
    [
        "id",
        "title",
        "original_language",
        "original_title",
        "popularity",
        "release_date",
        "vote_average",
        "vote_count",
        "release_date_status",
        "genre_ids"
    ]
].values.tolist()
```

```python
print(len(data))
print(data[0])
```

```python
insert_query = """
INSERT INTO movies (
    id,
    title,
    original_language,
    original_title,
    popularity,
    release_date,
    vote_average,
    vote_count,
    release_date_status,
    genre_ids
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""
```

```python
cursor.executemany(insert_query, data)
conn.commit()

print("Data inserted successfully!")
```

```python
df["release_date"] = df["release_date"].where(
    df["release_date"].notna(),
    None
)
```

```python
data = []

for row in df.itertuples(index=False):
    data.append((
        int(row.id),
        row.title,
        row.original_language,
        row.original_title,
        float(row.popularity),
        row.release_date.date() if pd.notna(row.release_date) else None,
        float(row.vote_average),
        int(row.vote_count),
        row.release_date_status,
        row.genre_ids
    ))
```

```python
print(len(data))
print(data[0])
```

```python
cursor.executemany(insert_query, data)
conn.commit()

print("Data inserted successfully!")
```
